### Discount Factor

- Future rewards are usually discounted.

- Where:
    γ = Discount Factor
    0 ≤ γ ≤ 1

- Interpretation
    γ = 0

- Only immediate reward matters
    γ = 1

- Future rewards matter equally
    Common values:
    0.95
    0.99

### Value Function

    Expected reward from a state.

    Meaning:

    How good is this state?
### Action Value Function (Q Function)

    Most important RL formula.

    Meaning:

    How good is taking action a
    in state s?
### Bellman Equation
    Foundation of RL.

    Interpretation:

    Current Value
    =
    Immediate Reward
    +
    Future Value

    Almost every RL algorithm is based on Bellman equations.

### Markov Decision Process (MDP)

    RL problems are modeled as MDPs.

    Components:

    S = States
    A = Actions
    P = Transition Probability
    R = Rewards
    γ = Discount Factor

    Written as:

    (S, A, P, R, γ)
## Types of RL Algorithms
### Model-Based RL

    Agent learns environment model.

    Example:

    Learn world first
    Then act

    Used less often.

### Model-Free RL

    Agent learns directly.

    Most popular.

    Examples:

    Q-Learning
    SARSA
    DQN
    PPO
### Value-Based Methods

    Learn state values.

    Examples:

    Q-Learning
    DQN
### Policy-Based Methods

    Learn policy directly.

    Examples:

    REINFORCE
    PPO
### Actor-Critic Methods

    Modern state-of-the-art.

    Examples:

    A2C
    A3C
    PPO

    Used heavily today.

### Q-Learning

    Most important beginner RL algorithm.

    Stores values in a Q-table.

    Example:

    State	Left	Right
    S1	1.2	5.6
    S2	3.4	1.1

    Choose action with highest Q-value.
